```python
import requests
from databricks.sdk import WorkspaceClient

APP_NAME   = "region-signal"
MAILER_URL = "https://script.google.com/macros/s/AKfycbydL3ajCHec_0suJmTkweJfrVr_w0YCTmIHXC7adae21gPb3DBRiizvlbPdcY8yBglC/exec"
SECRET     = "change-this-to-a-long-random-text"
RECIPIENTS = "vinoth.shankar@tigeranalytics.com"

w = WorkspaceClient()
w.apps.stop(APP_NAME).result()

html = """<p>Hello,</p>
<p>The <b>Pepsico Forecast Engine - Market Signal</b> app has been stopped and the 3-hour window is closed. Updates are no longer possible.</p>

<p>If a correction is needed, contact forecast_region_admin@pepsico.com</p>
<p>Thank you,<br>Forecast Engine - Market Signal</p>"""

r = requests.post(MAILER_URL, json={"secret": SECRET, "to": RECIPIENTS,
    "subject": "Pepsico Forecast Engine - Market Signal is now closed", "html": html}, timeout=30)
print(r.status_code, r.text[:200])
assert r.text.strip() == "sent", "Email was not sent: " + r.text[:200]
```
